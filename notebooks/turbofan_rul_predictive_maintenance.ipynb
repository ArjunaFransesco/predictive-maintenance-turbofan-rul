### 1. Ingest Sensor Telemetry
Load jet engine temperature, pressure, rotational speeds, and vibration signals.


```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('../data/raw/turbofan_sensor_telemetry.csv')
print(f'Total telemetry cycles: {len(df)}')
df.head()
```

### 2. Sensor Degradation Trajectory Visualization


```python
engine_1 = df[df['engine_id'] == 1]
fig, axes = plt.subplots(2, 2, figsize=(12, 6))
axes[0, 0].plot(engine_1['flight_cycle'], engine_1['sensor_T30'], color='crimson')
axes[0, 0].set_title('T30 HPC Outlet Temp Degradation')
axes[0, 1].plot(engine_1['flight_cycle'], engine_1['sensor_P30'], color='teal')
axes[0, 1].set_title('P30 Total Pressure Degradation')
axes[1, 0].plot(engine_1['flight_cycle'], engine_1['vibration_rms'], color='purple')
axes[1, 0].set_title('Vibration RMS Degradation')
axes[1, 1].plot(engine_1['flight_cycle'], engine_1['RUL_cycles'], color='darkgreen')
axes[1, 1].set_title('Ground Truth RUL')
plt.tight_layout()
plt.show()
```
